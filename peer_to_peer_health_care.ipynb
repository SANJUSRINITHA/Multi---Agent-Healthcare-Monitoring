```python
# ==============================================================
# MULTI-AGENT HEALTHCARE MONITORING SYSTEM
# PEER-TO-PEER (P2P) MULTI-AGENT ARCHITECTURE
# USING OPENAI API
# GOOGLE COLAB
# ==============================================================


# ==============================================================
# 0. INSTALL REQUIRED LIBRARY
# ==============================================================

!pip install -q openai


# ==============================================================
# 1. IMPORT LIBRARIES
# ==============================================================

import json
from openai import OpenAI


# ==============================================================
# 2. OPENAI API CONFIGURATION
# ==============================================================

print("=" * 70)
print("       OPENAI API CONFIGURATION")
print("=" * 70)

API_KEY = input("\nEnter your OpenAI API Key: ").strip()

client = OpenAI(api_key=API_KEY)

# Use a model available to your API account.
MODEL = "gpt-5.6"

print("\nOpenAI client initialized successfully.")


# ==============================================================
# 3. TITLE
# ==============================================================

print("\n")
print("=" * 70)
print("       MULTI-AGENT HEALTHCARE MONITORING SYSTEM")
print("              PEER-TO-PEER ARCHITECTURE")
print("                    OPENAI POWERED")
print("=" * 70)


# ==============================================================
# 4. PATIENT DATA
# ==============================================================

patient = {

    "Patient_ID": "P001",

    "Name": "Patient A",

    "Heart_Rate": 112,          # BPM

    "SpO2": 93,                 # %

    "Temperature": 38.2,        # °C

    "Blood_Pressure": "145/92"   # mmHg

}


print("\nPATIENT DATA")
print("-" * 70)

for key, value in patient.items():

    print(f"{key:20}: {value}")


# ==============================================================
# 5. OPENAI FUNCTION
# ==============================================================

def ask_openai(prompt):

    """
    Sends a prompt to the OpenAI API and returns
    the generated response.
    """

    try:

        response = client.responses.create(

            model=MODEL,

            input=prompt

        )

        return response.output_text

    except Exception as error:

        print("\nOPENAI API ERROR")
        print(error)

        return "OpenAI analysis unavailable."


# ==============================================================
# 6. BASE HEALTHCARE AGENT
# ==============================================================

class HealthcareAgent:

    def __init__(self, name):

        self.name = name

        self.messages = []


    # ----------------------------------------------------------
    # RECEIVE MESSAGE
    # ----------------------------------------------------------

    def receive_message(self, sender, message):

        self.messages.append(

            {

                "sender": sender,

                "message": message

            }

        )


    # ----------------------------------------------------------
    # SEND P2P MESSAGE
    # ----------------------------------------------------------

    def send_message(self, receiver, message):

        print("\n[P2P MESSAGE]")

        print(

            f"{self.name}  --->  "

            f"{receiver.name}"

        )

        print(f"Message: {message}")


        receiver.receive_message(

            self.name,

            message

        )


# ==============================================================
# 7. HEART RATE AGENT
# ==============================================================

class HeartAgent(HealthcareAgent):

    def analyze(self, heart_rate):

        # ------------------------------------------------------
        # Deterministic classification
        # ------------------------------------------------------

        if heart_rate < 60:

            status = "LOW"

        elif heart_rate <= 100:

            status = "NORMAL"

        else:

            status = "HIGH"


        # ------------------------------------------------------
        # OpenAI analysis
        # ------------------------------------------------------

        prompt = f"""

You are the Heart Rate Agent in a multi-agent
healthcare monitoring demonstration.

Patient Heart Rate:

{heart_rate} BPM

Use the following predefined classification:

Below 60 BPM = LOW
60 to 100 BPM = NORMAL
Above 100 BPM = HIGH

Provide:

1. Status
2. Short explanation
3. Whether continued monitoring is recommended

Do not diagnose a disease.

"""

        ai_analysis = ask_openai(prompt)


        # ------------------------------------------------------
        # Display
        # ------------------------------------------------------

        print("\n")
        print("-" * 70)
        print("❤️ HEART RATE AGENT")
        print("-" * 70)

        print(f"Heart Rate : {heart_rate} BPM")

        print(f"Status     : {status}")

        print("\nOpenAI Analysis:")

        print(ai_analysis)


        # ------------------------------------------------------
        # Result
        # ------------------------------------------------------

        return {

            "parameter": "Heart Rate",

            "value": heart_rate,

            "unit": "BPM",

            "status": status,

            "ai_analysis": ai_analysis

        }


# ==============================================================
# 8. OXYGEN AGENT
# ==============================================================

class OxygenAgent(HealthcareAgent):

    def analyze(self, spo2):

        # ------------------------------------------------------
        # Deterministic classification
        # ------------------------------------------------------

        if spo2 < 90:

            status = "CRITICAL"

        elif spo2 < 95:

            status = "LOW"

        else:

            status = "NORMAL"


        # ------------------------------------------------------
        # OpenAI analysis
        # ------------------------------------------------------

        prompt = f"""

You are the Oxygen Monitoring Agent in a multi-agent
healthcare monitoring demonstration.

Patient SpO2:

{spo2} %

Use the following predefined classification:

Below 90% = CRITICAL
90% to below 95% = LOW
95% or above = NORMAL

Provide:

1. Status
2. Short explanation
3. Whether continued monitoring is recommended

Do not diagnose a disease.

"""

        ai_analysis = ask_openai(prompt)


        # ------------------------------------------------------
        # Display
        # ------------------------------------------------------

        print("\n")
        print("-" * 70)
        print("🫁 OXYGEN AGENT")
        print("-" * 70)

        print(f"SpO2       : {spo2}%")

        print(f"Status     : {status}")

        print("\nOpenAI Analysis:")

        print(ai_analysis)


        # ------------------------------------------------------
        # Result
        # ------------------------------------------------------

        return {

            "parameter": "SpO2",

            "value": spo2,

            "unit": "%",

            "status": status,

            "ai_analysis": ai_analysis

        }


# ==============================================================
# 9. TEMPERATURE AGENT
# ==============================================================

class TemperatureAgent(HealthcareAgent):

    def analyze(self, temperature):

        # ------------------------------------------------------
        # Deterministic classification
        # ------------------------------------------------------

        if temperature < 36:

            status = "LOW"

        elif temperature <= 37.5:

            status = "NORMAL"

        elif temperature <= 39:

            status = "HIGH"

        else:

            status = "CRITICAL"


        # ------------------------------------------------------
        # OpenAI analysis
        # ------------------------------------------------------

        prompt = f"""

You are the Temperature Monitoring Agent in a
multi-agent healthcare monitoring demonstration.

Patient Temperature:

{temperature} °C

Use the following predefined classification:

Below 36°C = LOW
36°C to 37.5°C = NORMAL
Above 37.5°C to 39°C = HIGH
Above 39°C = CRITICAL

Provide:

1. Status
2. Short explanation
3. Whether continued monitoring is recommended

Do not diagnose a disease.

"""

        ai_analysis = ask_openai(prompt)


        # ------------------------------------------------------
        # Display
        # ------------------------------------------------------

        print("\n")
        print("-" * 70)
        print("🌡️ TEMPERATURE AGENT")
        print("-" * 70)

        print(f"Temperature : {temperature} °C")

        print(f"Status      : {status}")

        print("\nOpenAI Analysis:")

        print(ai_analysis)


        # ------------------------------------------------------
        # Result
        # ------------------------------------------------------

        return {

            "parameter": "Temperature",

            "value": temperature,

            "unit": "°C",

            "status": status,

            "ai_analysis": ai_analysis

        }


# ==============================================================
# 10. BLOOD PRESSURE AGENT
# ==============================================================

class BloodPressureAgent(HealthcareAgent):

    def analyze(self, blood_pressure):

        # ------------------------------------------------------
        # Split blood pressure
        # ------------------------------------------------------

        systolic, diastolic = map(

            int,

            blood_pressure.split("/")

        )


        # ------------------------------------------------------
        # Deterministic classification
        # ------------------------------------------------------

        if (

            systolic >= 140

            or

            diastolic >= 90

        ):

            status = "HIGH"


        elif (

            systolic < 90

            or

            diastolic < 60

        ):

            status = "LOW"


        else:

            status = "NORMAL"


        # ------------------------------------------------------
        # OpenAI analysis
        # ------------------------------------------------------

        prompt = f"""

You are the Blood Pressure Monitoring Agent
in a multi-agent healthcare monitoring demonstration.

Patient Blood Pressure:

{blood_pressure} mmHg

Use the following predefined classification:

Systolic >= 140 OR Diastolic >= 90 = HIGH

Systolic < 90 OR Diastolic < 60 = LOW

Otherwise = NORMAL

Provide:

1. Status
2. Short explanation
3. Whether continued monitoring is recommended

Do not diagnose a disease.

"""

        ai_analysis = ask_openai(prompt)


        # ------------------------------------------------------
        # Display
        # ------------------------------------------------------

        print("\n")
        print("-" * 70)
        print("🩸 BLOOD PRESSURE AGENT")
        print("-" * 70)

        print(

            f"Blood Pressure : "

            f"{blood_pressure} mmHg"

        )

        print(f"Status         : {status}")

        print("\nOpenAI Analysis:")

        print(ai_analysis)


        # ------------------------------------------------------
        # Result
        # ------------------------------------------------------

        return {

            "parameter": "Blood Pressure",

            "value": blood_pressure,

            "unit": "mmHg",

            "status": status,

            "ai_analysis": ai_analysis

        }


# ==============================================================
# 11. DECISION AGENT
# ==============================================================

class DecisionAgent(HealthcareAgent):

    def make_decision(self, results):

        print("\n")

        print("=" * 70)

        print("                    DECISION AGENT")

        print("=" * 70)


        # ------------------------------------------------------
        # Find abnormal parameters
        # ------------------------------------------------------

        abnormal = []


        for result in results:

            if result["status"] in [

                "HIGH",

                "LOW",

                "CRITICAL"

            ]:

                abnormal.append(result)


        # ------------------------------------------------------
        # Deterministic overall decision
        # ------------------------------------------------------

        if any(

            result["status"] == "CRITICAL"

            for result in results

        ):

            deterministic_status = "CRITICAL"


        elif len(abnormal) >= 2:

            deterministic_status = "WARNING"


        elif len(abnormal) == 1:

            deterministic_status = "WARNING"


        else:

            deterministic_status = "NORMAL"


        # ------------------------------------------------------
        # Prepare information for OpenAI
        # ------------------------------------------------------

        patient_information = {

            "Patient_ID": patient["Patient_ID"],

            "Heart_Rate": patient["Heart_Rate"],

            "SpO2": patient["SpO2"],

            "Temperature": patient["Temperature"],

            "Blood_Pressure": patient["Blood_Pressure"],

            "Agent_Results": results

        }


        # ------------------------------------------------------
        # OpenAI Decision Prompt
        # ------------------------------------------------------

        prompt = f"""

You are the Decision Agent in a P2P multi-agent
healthcare monitoring demonstration.

Several independent agents have analyzed the
patient's vital signs.

PATIENT INFORMATION:

{json.dumps(patient_information, indent=2)}


Analyze all the information together.

Determine:

1. Overall patient monitoring status
2. Abnormal parameters
3. Main concern
4. Recommended next monitoring/action step

Allowed overall statuses:

NORMAL
WARNING
CRITICAL


Important:

This is a software demonstration and not a medical
diagnosis. Do not claim to diagnose a disease.

Give a concise response in this format:

Overall Status:
Abnormal Parameters:
Assessment:
Recommendation:

"""


        # ------------------------------------------------------
        # Send to OpenAI
        # ------------------------------------------------------

        ai_decision = ask_openai(prompt)


        # ------------------------------------------------------
        # Display OpenAI Decision
        # ------------------------------------------------------

        print("\n🤖 OPENAI DECISION ANALYSIS")

        print("-" * 70)

        print(ai_decision)


        # ------------------------------------------------------
        # Display deterministic status
        # ------------------------------------------------------

        print("\nDeterministic Safety Status:")

        print(deterministic_status)


        return deterministic_status, ai_decision


# ==============================================================
# 12. CREATE AGENTS
# ==============================================================

print("\n")

print("=" * 70)

print("                 CREATING AGENTS")

print("=" * 70)


heart_agent = HeartAgent(

    "Heart Agent"

)


oxygen_agent = OxygenAgent(

    "Oxygen Agent"

)


temperature_agent = TemperatureAgent(

    "Temperature Agent"

)


bp_agent = BloodPressureAgent(

    "Blood Pressure Agent"

)


decision_agent = DecisionAgent(

    "Decision Agent"

)


print("\nAgents created successfully.")

print("\nAgents:")

print("1. Heart Agent")

print("2. Oxygen Agent")

print("3. Temperature Agent")

print("4. Blood Pressure Agent")

print("5. Decision Agent")


# ==============================================================
# 13. AGENT ANALYSIS
# ==============================================================

print("\n")

print("=" * 70)

print("                    AGENT ANALYSIS")

print("=" * 70)


heart_result = heart_agent.analyze(

    patient["Heart_Rate"]

)


oxygen_result = oxygen_agent.analyze(

    patient["SpO2"]

)


temperature_result = temperature_agent.analyze(

    patient["Temperature"]

)


bp_result = bp_agent.analyze(

    patient["Blood_Pressure"]

)


# ==============================================================
# 14. PEER-TO-PEER COMMUNICATION
# ==============================================================

print("\n")

print("=" * 70)

print("              PEER-TO-PEER COMMUNICATION")

print("=" * 70)


# --------------------------------------------------------------
# Heart Agent → Oxygen Agent
# --------------------------------------------------------------

heart_agent.send_message(

    oxygen_agent,

    {

        "parameter": "Heart Rate",

        "status": heart_result["status"],

        "value": heart_result["value"]

    }

)


# --------------------------------------------------------------
# Oxygen Agent → Temperature Agent
# --------------------------------------------------------------

oxygen_agent.send_message(

    temperature_agent,

    {

        "parameter": "SpO2",

        "status": oxygen_result["status"],

        "value": oxygen_result["value"]

    }

)


# --------------------------------------------------------------
# Temperature Agent → Blood Pressure Agent
# --------------------------------------------------------------

temperature_agent.send_message(

    bp_agent,

    {

        "parameter": "Temperature",

        "status": temperature_result["status"],

        "value": temperature_result["value"]

    }

)


# --------------------------------------------------------------
# Blood Pressure Agent → Heart Agent
# --------------------------------------------------------------

bp_agent.send_message(

    heart_agent,

    {

        "parameter": "Blood Pressure",

        "status": bp_result["status"],

        "value": bp_result["value"]

    }

)


# ==============================================================
# 15. INFORMATION SHARING WITH DECISION AGENT
# ==============================================================

print("\n")

print("=" * 70)

print("             INFORMATION SHARING")

print("=" * 70)


heart_agent.send_message(

    decision_agent,

    heart_result

)


oxygen_agent.send_message(

    decision_agent,

    oxygen_result

)


temperature_agent.send_message(

    decision_agent,

    temperature_result

)


bp_agent.send_message(

    decision_agent,

    bp_result

)


# ==============================================================
# 16. COLLECT RESULTS
# ==============================================================

results = [

    heart_result,

    oxygen_result,

    temperature_result,

    bp_result

]


# ==============================================================
# 17. FINAL DECISION
# ==============================================================

overall_status, ai_decision = (

    decision_agent.make_decision(

        results

    )

)


# ==============================================================
# 18. FINAL REPORT
# ==============================================================

print("\n")

print("=" * 70)

print("                       FINAL REPORT")

print("=" * 70)


print(

    f"\nPatient ID       : "

    f"{patient['Patient_ID']}"

)


print(

    f"Patient Name     : "

    f"{patient['Name']}"

)


print("\n")

print("MONITORING RESULTS")

print("-" * 70)


print(

    f"{'Parameter':20}"

    f"{'Value':15}"

    f"{'Status':15}"

)


print("-" * 70)


for result in results:

    value_with_unit = (

        str(result["value"])

        + " "

        + result["unit"]

    )

    print(

        f"{result['parameter']:20}"

        f"{value_with_unit:15}"

        f"{result['status']:15}"

    )


print("-" * 70)


print("\n")

print(

    f"OVERALL STATUS : "

    f"{overall_status}"

)


print("\n")

print("OPENAI DECISION AGENT OUTPUT")

print("-" * 70)

print(ai_decision)


# ==============================================================
# 19. P2P MESSAGE SUMMARY
# ==============================================================

print("\n")

print("=" * 70)

print("                  P2P MESSAGE SUMMARY")

print("=" * 70)


print("\nMessages received by Heart Agent:")

for message in heart_agent.messages:

    print(message)


print("\nMessages received by Oxygen Agent:")

for message in oxygen_agent.messages:

    print(message)


print("\nMessages received by Temperature Agent:")

for message in temperature_agent.messages:

    print(message)


print("\nMessages received by Blood Pressure Agent:")

for message in bp_agent.messages:

    print(message)


print("\nMessages received by Decision Agent:")

for message in decision_agent.messages:

    print(message)


# ==============================================================
# 20. SYSTEM COMPLETION
# ==============================================================

print("\n")

print("=" * 70)

print("        P2P MULTI-AGENT HEALTHCARE SYSTEM")

print("                    COMPLETED")

print("=" * 70)

print("\nArchitecture:")

print("Patient Data")

print("     ↓")

print("Heart Agent")

print("     ↕")

print("Oxygen Agent")

print("     ↕")

print("Temperature Agent")

print("     ↕")

print("Blood Pressure Agent")

print("     ↓")

print("Decision Agent")

print("     ↓")

print("OpenAI API")

print("     ↓")

print("Final Patient Monitoring Report")

print("\n")

print("NOTE:")

print(

    "This system is a healthcare monitoring "

    "demonstration and is not a medical diagnosis tool."

)

print("=" * 70)
```
